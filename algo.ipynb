```python
'''find the largest of three numbers
    example: a=10, b=25, c=15 '''
'''Algorithm:
    1. Start
    2. input three numbers a, b, c
    3. if a  is grater than b and c , print a
    4. otherwise if b is grater than a and c, print b
    5. otherwise print c
    6. Stop'''
    
'''Result is 25 (largest number)'''

# let create the program to find the largest of three numbers
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))
if a > b and a > c:
    print("The largest number is:", a)
elif b > a and b > c:
    print("The largest number is:", b)
else:
    print("The largest number is:", c)
    
```

```python
'''find the largest among N numbers
    example: a=10, b=25, c=15, d=30, e=5 '''
'''Algorithm:
    1. Start
    2. input N numbers
    3. set the first number as largest
    4. compare the largest with the next number
    5. if the next number is greater than largest, set it as largest
    6. repeat step 4 and 5 for all numbers
    7. print the largest number
    8. Stop'''
    # Let's create the program to find the largest among N numbers
example_numbers = [10, 25, 15, 30, 5]
largest = example_numbers[0]
for number in example_numbers:
    if number > largest:
        largest = number
print("The largest number among the given numbers is:", largest)
```

```python
'''find the smallest among N numbers
    example: a=10, b=25, c=15, d=30, e=5 '''
'''Algorithm:
    1. Start
    2. input N numbers
    3. set the first number as smallest
    4. compare the smallest with the next number
    5. if the next number is smaller than smallest, set it as smallest
    6. repeat step 4 and 5 for all numbers
    7. print the smallest number
    8. Stop'''
    # Let's create the program to find the smallest among N numbers
example_numbers = [10, 25, 15, 30, 5]
smallest = example_numbers[0]
for number in example_numbers:
    if number< smallest:
        smallest = number
print("The smallest number among the given numbers is:", smallest)
```
